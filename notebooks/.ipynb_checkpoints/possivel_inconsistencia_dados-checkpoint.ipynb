### Desconfiança de inconsistencia de dados, 3 estados Faltando em 2023
##### Logo, estes estados não aparecem no treino mas aparecem no teste 2024

```python
uf_treino = set(X_train["uf"].unique())
uf_teste = set(X_test["uf"].unique())

print("UFs no teste mas não no treino:", uf_teste - uf_treino)
print("UFs no treino mas não no teste:", uf_treino - uf_teste)

# Quantos registros são afetados
registros_afetados = X_test[~X_test["uf"].isin(uf_treino)]
print(f"Registros afetados: {len(registros_afetados):,} de {len(X_test):,}")
```

```python
uf_2023 = set(df_preparado[df_preparado["ano"] == 2023]["uf"].unique())
uf_2024 = set(df_preparado[df_preparado["ano"] == 2024]["uf"].unique())

print("Só em 2024:", uf_2024 - uf_2023)
print("Só em 2023:", uf_2023 - uf_2024)

print("\nTotal de UFs distintas em 2023:", len(uf_2023))
print("Total de UFs distintas em 2024:", len(uf_2024))
```

```python
def normalizar(texto):
    if pd.isna(texto):
        return texto
    texto = unicodedata.normalize("NFKD", texto).encode("ASCII", "ignore").decode("utf-8")
    return texto.strip().lower()

df_preparado["uf_normalizado"] = df_preparado["uf"].apply(normalizar)

uf_2023_norm = set(df_preparado[df_preparado["ano"] == 2023]["uf_normalizado"].unique())
uf_2024_norm = set(df_preparado[df_preparado["ano"] == 2024]["uf_normalizado"].unique())

print("Diferença após normalizar (só 2024):", uf_2024_norm - uf_2023_norm)
print("Diferença após normalizar (só 2023):", uf_2023_norm - uf_2024_norm)
```

```python
## Olhando para o dataset original pra ver se o problema já exisita lá

CAMINHO_ALUNOS = r"E:\POS_GRADUACAO\FASE3\Bases\alfabetizacao\alunos.csv"

con = duckdb.connect()

resultado = con.execute(f"""
    SELECT
        ano,
        CASE
            WHEN CAST(id_municipio AS VARCHAR) LIKE '12%' THEN 'Acre'
            WHEN CAST(id_municipio AS VARCHAR) LIKE '35%' THEN 'São Paulo'
            WHEN CAST(id_municipio AS VARCHAR) LIKE '53%' THEN 'Distrito Federal'
        END AS estado,
        COUNT(*) AS registros
    FROM read_csv_auto('{CAMINHO_ALUNOS}')
    WHERE CAST(id_municipio AS VARCHAR) LIKE '12%'
       OR CAST(id_municipio AS VARCHAR) LIKE '35%'
       OR CAST(id_municipio AS VARCHAR) LIKE '53%'
    GROUP BY ano, estado
    ORDER BY estado, ano
""").fetchall()

for ano, estado, registros in resultado:
    print(f"{estado} | {ano}: {registros:,} registros")
```

### Comprovado que o DataSet original não continham dados para esses 3 estados em 2023
